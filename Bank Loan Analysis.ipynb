```python
# Calculate Funded Amount

df["Funded_Amount"] = df.apply(
    lambda row: row["Loan_Amount"]
    if row["Application_Status"] == "Approved"
    else 0,
    axis=1
)

# Calculate Received Amount
df["Received_Amount"] = df.apply(
    lambda row: (
        row["Loan_Amount"] * 1.10
        if row["Repayment_Status"] == "Paid"
        else row["Loan_Amount"] * 0.90
        if row["Repayment_Status"] == "Late"
        else row["Loan_Amount"] * 0.40
        if row["Repayment_Status"] == "Default"
        else 0
    ),
    axis=1
)

print("Funded and Received Amounts Created!")
print()

print(
    df[
        [
            "Loan_ID",
            "Loan_Amount",
            "Application_Status",
            "Repayment_Status",
            "Funded_Amount",
            "Received_Amount"
        ]
    ].head(10)
)
```

```python
# Calculate Funded Amount

df["Funded_Amount"] = df.apply(
    lambda row: row["Loan_Amount"]
    if row["Application_Status"] == "Approved"
    else 0,
    axis=1
)

# Calculate Received Amount
df["Received_Amount"] = df.apply(
    lambda row: (
        row["Loan_Amount"] * 1.10
        if row["Repayment_Status"] == "Paid"
        else row["Loan_Amount"] * 0.90
        if row["Repayment_Status"] == "Late"
        else row["Loan_Amount"] * 0.40
        if row["Repayment_Status"] == "Default"
        else 0
    ),
    axis=1
)

print("Funded and Received Amounts Created!")
print()

print(
    df[
        [
            "Loan_ID",
            "Loan_Amount",
            "Application_Status",
            "Repayment_Status",
            "Funded_Amount",
            "Received_Amount"
        ]
    ].head(10)
)
```

```python
# Classify loans into Good, Bad, and Rejected

df["Loan_Quality"] = df.apply(
    lambda row: "Good Loan"
    if row["Application_Status"] == "Approved"
    and row["Repayment_Status"] == "Paid"
    else "Bad Loan"
    if row["Application_Status"] == "Approved"
    and row["Repayment_Status"] in ["Late", "Default"]
    else "Rejected",
    axis=1
)

print("Loan classification completed!")
print()

print(
    df[
        [
            "Loan_ID",
            "Customer_Name",
            "Application_Status",
            "Repayment_Status",
            "Loan_Quality"
        ]
    ]
)
```

```python
# Create time-based columns

df["Application_Year"] = df["Application_Date"].dt.year

df["Application_Month"] = df["Application_Date"].dt.month_name()

df["Application_Month_Number"] = df["Application_Date"].dt.month

df["Year_Month"] = df["Application_Date"].dt.strftime("%Y-%m")

print("Time columns created successfully!")
print()

print(
    df[
        [
            "Loan_ID",
            "Application_Date",
            "Application_Year",
            "Application_Month",
            "Year_Month"
        ]
    ].head(10)
)
```

```python
# Bank Loan KPIs

total_applications = len(df)

approved_loans = (df["Application_Status"] == "Approved").sum()

rejected_loans = (df["Application_Status"] == "Rejected").sum()

total_funded_amount = df["Funded_Amount"].sum()

total_received_amount = df["Received_Amount"].sum()

approval_rate = (approved_loans / total_applications) * 100

good_loans = (df["Loan_Quality"] == "Good Loan").sum()

bad_loans = (df["Loan_Quality"] == "Bad Loan").sum()

bad_loan_rate = (bad_loans / approved_loans) * 100


print("BANK LOAN KPI SUMMARY")
print("=====================")

print("Total Applications :", total_applications)
print("Approved Loans     :", approved_loans)
print("Rejected Loans     :", rejected_loans)
print("Total Funded Amount: ₹", total_funded_amount)
print("Total Received Amt : ₹", round(total_received_amount, 2))
print("Approval Rate      :", round(approval_rate, 2), "%")
print("Good Loans         :", good_loans)
print("Bad Loans          :", bad_loans)
print("Bad Loan Rate      :", round(bad_loan_rate, 2), "%")
```

```python

import pandas as pd

df = pd.read_csv("Bank_Loan_Analysis_Dataset.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# ---------------- PREPARE DATA ----------------
df = df.copy()
df.columns = df.columns.str.strip()

required = [
    "Loan_ID", "Region", "Application_Status", "Repayment_Status",
    "Application_Date", "Funded_Amount", "Received_Amount",
    "Loan_Quality", "Application_Year", "Employment_Type",
    "Interest_Rate", "Annual_Income", "Loan_Amount"
]

missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

for c in [
    "Funded_Amount", "Received_Amount", "Loan_Amount",
    "Annual_Income", "Interest_Rate"
]:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df["Application_Date"] = pd.to_datetime(
    df["Application_Date"], errors="coerce"
)

# ---------------- SIMPLE THEME ----------------
BG = "#F5F6F7"
TEXT = "#30343B"
MUTED = "#737982"
BORDER = "#E0E3E7"
BLUE = "#6F8FA8"
GREEN = "#7D9887"
RED = "#B98484"
GREY = "#A7ADB4"

display(HTML(f"""
<style>
.jp-OutputArea-output {{ background-color: {BG}; }}
</style>
<div style="background:white;color:{TEXT};padding:22px 25px;
border-bottom:2px solid {BORDER};font-family:Arial">
  <div style="font-size:11px;letter-spacing:1.5px;color:{MUTED}">
    FINANCIAL PERFORMANCE REPORT
  </div>
  <h1 style="margin:8px 0;font-size:27px;font-weight:600">
    BANK LOAN ANALYSIS
  </h1>
  <p style="margin:0;color:{MUTED}">
    Loan Applications | Funding | Repayment | Credit Risk
  </p>
</div>
"""))

# ---------------- FILTERS ----------------
def get_options(col):
    return ["All"] + sorted(df[col].dropna().astype(str).unique().tolist())

region_filter = widgets.Dropdown(
    options=get_options("Region"),
    description="Region:",
    layout=widgets.Layout(width="220px")
)

year_filter = widgets.Dropdown(
    options=["All"] + sorted(
        df["Application_Year"].dropna().astype(str).unique().tolist()
    ),
    description="Year:",
    layout=widgets.Layout(width="200px")
)

quality_filter = widgets.Dropdown(
    options=get_options("Loan_Quality"),
    description="Loan Quality:",
    layout=widgets.Layout(width="260px")
)

output = widgets.Output()

# ---------------- KPI CARD ----------------
def make_card(label, value, accent):
    return f"""
    <div style="background:white;padding:17px 19px;
    border:1px solid {BORDER};border-top:3px solid {accent};
    border-radius:5px;flex:1;min-width:135px;font-family:Arial">
      <div style="font-size:11px;color:{MUTED};font-weight:500">
        {label}
      </div>
      <div style="font-size:23px;color:{TEXT};
      font-weight:600;margin-top:10px">{value}</div>
    </div>
    """

# ---------------- CHART STYLE ----------------
def style_chart(fig):
    fig.update_layout(
        template="plotly_white",
        paper_bgcolor="white",
        plot_bgcolor="white",
        font=dict(family="Arial", color=TEXT, size=12),
        title_font=dict(size=16, color=TEXT),
        margin=dict(l=35, r=25, t=65, b=40),
        colorway=[BLUE, GREEN, GREY, RED],
        legend=dict(
            orientation="h", yanchor="bottom", y=1.02,
            xanchor="right", x=1
        )
    )
    fig.update_xaxes(showgrid=False, linecolor=BORDER)
    fig.update_yaxes(gridcolor="#ECEEF0", zeroline=False)
    return fig

def show_chart(fig):
    fig = style_chart(fig)
    fig.show()

# ---------------- BUILD DASHBOARD ----------------
def update_dashboard(change=None):
    data = df.copy()

    if region_filter.value != "All":
        data = data[data["Region"].astype(str) == region_filter.value]

    if year_filter.value != "All":
        data = data[
            data["Application_Year"].astype(str) == year_filter.value
        ]

    if quality_filter.value != "All":
        data = data[
            data["Loan_Quality"].astype(str) == quality_filter.value
        ]

    with output:
        clear_output(wait=True)

        if data.empty:
            print("No records match the selected filters.")
            return

        total = len(data)
        funded = data["Funded_Amount"].sum()
        received = data["Received_Amount"].sum()
        approved = data["Application_Status"].astype(str).str.lower().eq("approved").sum()
        good = data["Loan_Quality"].astype(str).str.lower().eq("good loan").sum()
        bad = data["Loan_Quality"].astype(str).str.lower().eq("bad loan").sum()

        # KPI row
        kpis = [
            ("TOTAL APPLICATIONS", f"{total:,}", BLUE),
            ("FUNDED AMOUNT", f"₹{funded:,.0f}", GREY),
            ("AMOUNT RECEIVED", f"₹{received:,.0f}", GREEN),
            ("APPROVAL RATE", f"{approved/total:.1%}", BLUE),
            ("GOOD LOANS", f"{good:,}", GREEN),
            ("BAD LOANS", f"{bad:,}", RED)
        ]

        display(HTML(
            '<div style="display:flex;flex-wrap:wrap;gap:12px;margin:14px 0 24px">'
            + "".join(make_card(*item) for item in kpis)
            + "</div>"
        ))

        # Good vs bad loans
        quality = data["Loan_Quality"].value_counts().reset_index()
        quality.columns = ["Loan Quality", "Applications"]

        fig = px.pie(
            quality, names="Loan Quality", values="Applications",
            title="Loan Quality Distribution", hole=0.55,
            color_discrete_sequence=[GREEN, RED, GREY]
        )
        fig.update_traces(textposition="inside", textinfo="percent+label")
        show_chart(fig)

        # Application status
        status = data["Application_Status"].value_counts().reset_index()
        status.columns = ["Application Status", "Applications"]

        fig = px.bar(
            status, x="Application Status", y="Applications",
            title="Applications by Status", text="Applications",
            color_discrete_sequence=[BLUE]
        )
        show_chart(fig)

        # Monthly trends
        dated = data.dropna(subset=["Application_Date"]).copy()

        if not dated.empty:
            dated["Month"] = dated["Application_Date"].dt.to_period("M").astype(str)

            monthly = dated.groupby("Month", as_index=False).agg(
                Applications=("Loan_ID", "count"),
                Funded=("Funded_Amount", "sum"),
                Received=("Received_Amount", "sum")
            )

            fig = px.line(
                monthly, x="Month", y="Applications",
                title="Monthly Loan Applications", markers=True,
                color_discrete_sequence=[BLUE]
            )
            show_chart(fig)

            fig = go.Figure()
            fig.add_trace(go.Scatter(
                x=monthly["Month"], y=monthly["Funded"],
                name="Funded Amount", mode="lines+markers",
                line=dict(color=BLUE, width=2)
            ))
            fig.add_trace(go.Scatter(
                x=monthly["Month"], y=monthly["Received"],
                name="Received Amount", mode="lines+markers",
                line=dict(color=GREEN, width=2)
            ))
            fig.update_layout(title="Funded Amount vs Amount Received")
            show_chart(fig)

        # Regional performance
        regional = data.groupby("Region", as_index=False).agg(
            Applications=("Loan_ID", "count"),
            Funded=("Funded_Amount", "sum")
        ).sort_values("Funded", ascending=False)

        fig = px.bar(
            regional, x="Region", y="Funded",
            title="Funded Amount by Region",
            color_discrete_sequence=[BLUE]
        )
        show_chart(fig)

        # Repayment status
        repayment = data["Repayment_Status"].value_counts().reset_index()
        repayment.columns = ["Repayment Status", "Applications"]

        fig = px.bar(
            repayment, x="Repayment Status", y="Applications",
            title="Repayment Status Analysis", text="Applications",
            color_discrete_sequence=[GREEN]
        )
        show_chart(fig)

        # Employment and loan quality
        employment = data.groupby(
            ["Employment_Type", "Loan_Quality"]
        ).size().reset_index(name="Applications")

        fig = px.bar(
            employment, x="Employment_Type", y="Applications",
            color="Loan_Quality", barmode="group",
            title="Loan Quality by Employment Type",
            color_discrete_map={
                "Good Loan": GREEN,
                "Bad Loan": RED,
                "Rejected": GREY
            }
        )
        show_chart(fig)

        # Interest rate and loan quality
        fig = px.box(
            data, x="Loan_Quality", y="Interest_Rate",
            title="Interest Rate by Loan Quality",
            color_discrete_sequence=[BLUE]
        )
        show_chart(fig)

        # Income vs loan amount
        fig = px.scatter(
            data, x="Annual_Income", y="Loan_Amount",
            title="Annual Income vs Loan Amount",
            color="Loan_Quality",
            color_discrete_map={
                "Good Loan": GREEN,
                "Bad Loan": RED,
                "Rejected": GREY
            }
        )
        show_chart(fig)

        # Summary
        display(HTML(f"""
        <div style="margin-top:15px;padding:18px;background:white;
        border:1px solid {BORDER};border-radius:5px;
        font-family:Arial;color:{TEXT}">
          <b>REPORT SUMMARY</b><br><br>
          Applications analyzed: {total:,}<br>
          Approved applications: {approved:,}<br>
          Total funded: ₹{funded:,.2f}<br>
          Total received: ₹{received:,.2f}
        </div>
        """))

region_filter.observe(update_dashboard, names="value")
year_filter.observe(update_dashboard, names="value")
quality_filter.observe(update_dashboard, names="value")

display(widgets.HBox(
    [region_filter, year_filter, quality_filter],
    layout=widgets.Layout(gap="12px", margin="15px 0")
))
display(output)

update_dashboard()
```
